1) number guessing game

```python
import random as rd
tries=1
x=rd.randint(1,100)
flag=0
while flag!=1:
     num=int(input("Enter the number: "))
     if num>x:
          print("TOO HIGH")
          tries=tries+1
     elif num<x:
          print("TOO LOW")
          tries=tries+1
     else:
          print("YOU GOTCHA! in",tries,"try")
          flag=1
```

2) password generator

```python
import random as rd
import string
num=int(input("Enter the length of the password"))

p=[]
p.append(rd.choice(string.ascii_lowercase))
p.append(rd.choice(string.ascii_uppercase))
p.append(rd.choice(string.ascii_letters))
p.append(rd.choice(string.punctuation))

chars=string.ascii_lowercase+string.ascii_uppercase+string.punctuation+string.digits
p+=[rd.choice(chars) for i in range(num-5)]
rd.shuffle(p)
passw="".join(p)
print("Your password is",passw)


```

4) currency converter with dictionary

```python
import yfinance as yf

inr=yf.Ticker("USDINR=X").fast_info["last_price"]
yen=yf.Ticker("USDJPY=X").fast_info["last_price"]

rates={
     1: 1.00,
     2: inr,
     3: yen
}
symbols={
     1: 'usd',
     2: 'inr',
     3: 'yen'
}

print("1.usd,2.inr,3.yen")
c1=int(input("from(currency number): "))
c2=int(input("to(currency number): "))
amnt=float(input("Enter the amount: "))

usd_amnt=amnt/rates[c1]
res=usd_amnt*rates[c2]

print("The",amnt,symbols[c1],"=>",round(res,2),symbols[c2])
```

5) contact book

```python
#contact book

contacts={
     "sample": 987654321,
}
flag=0

def viewAll():
     print("All Saved Contacts Down Below - \n")
     print(contacts,'\n')

def search():
     op1=int(input("Type 1 to search by name\nType 2 to search by number: "))
     match op1:
          case 1: 
               name=input("Enter the name: ")
               print(name,"=>",contacts.get(name),"\n")
               return
          case 2:
               num=int(input("Enter the number: "))
               for key,value in contacts.items():
                    if num==value:
                         print("It is",key,"\n")
                         return

def add():
     name=input("Enter the name: ")
     num=int(input("Enter the number: "))
     contacts.update({name:num})

def delete():
     op1=int(input("Type 1 to delete by name\nType 2 to delete by number: "))
     match op1:
          case 1: 
               name=input("Enter the name: ")
               del contacts[name]
               print("\nDeleted successfully\n")
               return
          case 2:
               num=int(input("Enter the number: "))
               for key,value in contacts.items():
                    if num==value:
                         del contacts[key]
                         print("\nDeleted successfully\n")
                         return

while flag!=1:
     print("1.viewAll\n2.search\n3.add\n4.delete\npress any other digits to exit")
     op=int(input("Please enter the option number: "))
     match op:
          case 1: viewAll()
          case 2: search()
          case 3: add()
          case 4: delete()
          case _: 
                 print("Exited...")
                 flag=1

```

7. rock-paper-scissors

```python
import random as rd

#computer-Choice = x
#user-Choice = y

choices={
     1:"rock",
     2:"paper",
     3:"scissors"
}

rounds=0
Xp=0
Yp=0

print("ROCK_PAPER_SCISSORS\n\n""Winning Rules as follows:\nRock vs paper - paper wins\nRock vs scissor - Rock wins\npaper vs scissor - scissor wins")
print("TYPE:\n1-rock\n2-paper\n3-scissors\n\n")

rounds=int(input("Enter the rounds that you wanna play: "))

for i in range(0,rounds):
     print("\n\nRound:",i+1,"\n")
     x=rd.randint(1,3)
     y=int(input("Enter your choice: "))
     print("Your choice =",choices.get(y))
     print("Computer choice =",choices.get(x))
     if y==x+1:
          Yp=Yp+1
          print("You won")
     elif x==y+1:
          Xp=Xp+1
          print("You lost")
     elif x==y:
          print("equal! doesn't count")
     else:
          prior= True if y==x+2 else False
          match prior:
               case True:
                     Xp=Xp+1
                     print("You lost")
               case False: 
                     Yp=Yp+1
                     print("You won")
     print("\nSCORE BOARD: \nYou:",Yp,"\nBot:",Xp)

if Yp>Xp:
     print("CONGRATZ! YOU WON THE GAME\n")
elif Xp>Yp:
     print("HAHA! YOU LOST\n")
else:
     print("Somehow, you guys scored equal!\n")
```
